## Exploring data with PROCedures

Four procedures for a first look at any table: `PRINT`, `MEANS`, `UNIVARIATE`, and `FREQ`.

```{=html}
<div class="anno"><div class="anno-row"><pre class="anno-code"><span class="cmt">/* prints n rows for the listed columns */</span>
<span class="kw">proc print</span> data=<i>input-table</i>(obs=<i>n</i>);
    <span class="kw">var</span> <i>col-name(s)</i>;
<span class="kw">run</span>;

<span class="cmt">/* summary statistics for the listed columns */</span>
<span class="kw">proc means</span> data=<i>input-table</i>;
    <span class="kw">var</span> <i>col-name(s)</i>;
<span class="kw">run</span>;

<span class="cmt">/* summary statistics, with more about the distribution */</span>
<span class="kw">proc univariate</span> data=<i>input-table</i>;
    <span class="kw">var</span> <i>col-name(s)</i>;
<span class="kw">run</span>;

<span class="cmt">/* frequency table for each listed column */</span>
<span class="kw">proc freq</span> data=<i>input-table</i>;
    <span class="kw">tables</span> <i>col-name(s)</i>;
<span class="kw">run</span>;</pre></div></div>
```

## `PROC PRINT`

Lists the rows of a table. `(obs=10)` keeps the first 10 rows; `var` picks the columns.

````sas
/* print the first 10 observations of the sashelp.cars table */
proc print data=sashelp.cars (obs=10);
    var Make Model Type MSRP;
run;
````

::: {.note style="display:inline-block"}
**Default:** `proc print data=sashelp.cars; run;` prints the **entire table**, every row and every column (all 428 cars).
:::

## `PROC MEANS`

Summary statistics (N, mean, standard deviation, min, max) for numeric columns.

````sas
/* summary statistics for EngineSize and Horsepower */
proc means data=sashelp.cars;
    var EngineSize Horsepower;
run;
````

::: {.note style="display:inline-block"}
**Default:** without a `var` statement, `PROC MEANS` gives summary statistics for **every numeric** column.
:::

## Quick exercise

What happens if we limit the number of observations, e.g. `(obs=15)`?

<div class="solution-start"></div>

````sas
proc means data=sashelp.cars (obs=15);
    var EngineSize Horsepower;
run;
````

`N` drops to **15**: the statistics are computed from the **first 15 rows only**, not the whole table.

## `PROC UNIVARIATE`

Descriptive statistics with more about the **distribution**. We get **5 tables**:
Moments, Basic Statistical Measures, Tests for Location, Quantiles, and Extreme Observations.

````sas
/* descriptive statistics / distribution for MPG_Highway */
proc univariate data=sashelp.cars;
    var MPG_Highway;
run;
````

## `PROC FREQ`

Counts how often each value appears. Use it on **categorical** columns.

````sas
/* frequency table for a categorical column */
proc freq data=sashelp.cars;
    tables Origin;
run;
````

## ODS: the Output Delivery System

ODS controls the **type and format** of the output your SAS code produces: how SAS displays and saves results.

- `ods trace on;` writes the **name** of every output table to the **log**.
- `ods select` *table-name*`;` keeps **only** that table in the results.

**Exercise:** use `ods trace` and `ods select` to output **only** the Extreme Observations table for `Damage`. Start here, then check the log:

````sas
ods trace on;
proc univariate data=ma505.ow_stats;
    var Damage;
run;
````

<div class="solution-start"></div>

````sas
/* the log lists "Name: ExtremeObs" for the Extreme Observations table */
ods trace off;

ods select ExtremeObs;
proc univariate data=ma505.ow_stats;
    var Damage;
run;
````

## Filtering rows with `WHERE`

:::: {.columns}
::: {.column width="45%"}
```{=html}
<div class="syntax-box"><b>PROC</b> <i>procedure-name</i> … ;<br>
&nbsp;&nbsp;&nbsp;&nbsp;<b>WHERE</b> <i>expression</i>;<br>
<b>RUN;</b></div>
```
:::
::: {.column width="55%"}
Works in `PROC PRINT`, `MEANS`, `FREQ`, and `UNIVARIATE`: the procedure only uses rows where the expression is **true**.
:::
::::

::: {.fragment}
| Operator | Symbol | Mnemonic |
|---|:---:|:---:|
| Equals | `=` | `EQ` |
| Not equal | `^=` or `~=` | `NE` |
| Greater than | `>` | `GT` |
| Less than | `<` | `LT` |
| Greater than or equal to | `>=` | `GE` |
| Less than or equal to | `<=` | `LE` |
:::

## More `WHERE` operators

```{=html}
<div class="where-list">
<div class="fragment"><b>AND</b> / <b>OR</b>: combine expressions
<pre class="anno-code"><span class="kw">where</span> Type = "SUV" <span class="red-text">and</span> MSRP <= 30000;</pre></div>
<div class="fragment"><b>IN</b> / <b>NOT IN</b>: match any value in a list
<pre class="anno-code"><span class="kw">where</span> Type <span class="red-text">in</span> ("SUV", "Truck");</pre></div>
<div class="fragment"><b>IS MISSING</b> / <b>IS NOT MISSING</b>
<pre class="anno-code"><span class="kw">where</span> Cylinders <span class="red-text">is missing</span>;</pre></div>
<div class="fragment"><b>BETWEEN … AND</b>: an inclusive range
<pre class="anno-code"><span class="kw">where</span> MSRP <span class="red-text">between</span> 36000 <span class="red-text">and</span> 50000;</pre></div>
<div class="fragment"><b>Dates</b>: use a SAS date constant (Day 2)
<pre class="anno-code"><span class="kw">where</span> Date > "01NOV2016"<span class="red-text">d</span>;</pre></div>
</div>
```

::: {.fragment .red-text}
Character values are **case-sensitive**: `"suv"` does not match `"SUV"`.
:::

## `WHERE`

````sas
/* SUVs that cost $30,000 or less */
proc print data=sashelp.cars;
    where Type = "SUV" and MSRP <= 30000;
    var Make Model Type MSRP;
run;
````

## `WHERE` with a date

`sashelp.stocks` has a `Date` column, so we filter it with a date constant.

````sas
/* IBM's monthly stock prices in 2005 */
proc print data=sashelp.stocks;
    where Stock = "IBM" and Date >= "01JAN2005"d;
    var Stock Date Open Close;
run;
````

## SAS macro variables

A **macro variable** is a placeholder: a name that stores a value you can reuse throughout a program.
It makes code **dynamic, reusable, and flexible**: change the value once, and every place that uses it changes too.

```{=html}
<div class="macro-compare">
<div class="anno"><div class="anno-row"><pre class="anno-code"><span class="kw">proc print</span> data=sashelp.cars;
    <span class="kw">where</span> Make = "Kia";
    <span class="kw">var</span> Type Make Model MSRP;
<span class="kw">run</span>;

<span class="kw">proc freq</span> data=sashelp.cars;
    <span class="kw">where</span> Make = "Kia";
    <span class="kw">tables</span> Origin;
<span class="kw">run</span>;</pre></div></div>
<div class="macro-arrow fragment" data-fragment-index="1"><div class="arrow-right black"></div></div>
<div class="anno fragment" data-fragment-index="1"><div class="anno-row"><pre class="anno-code"><span class="cmt">/* set a macro variable */</span>
<span class="libref-circle"><span class="kw">%let</span> CarMake = Kia;</span>

<span class="kw">proc print</span> data=sashelp.cars;
    <span class="kw">where</span> Make = <span class="libref-circle">"&amp;CarMake"</span>;
    <span class="kw">var</span> Type Make Model MSRP;
<span class="kw">run</span>;

<span class="kw">proc freq</span> data=sashelp.cars;
    <span class="kw">where</span> Make = <span class="libref-circle">"&amp;CarMake"</span>;
    <span class="kw">tables</span> Origin;
<span class="kw">run</span>;</pre></div></div>
</div>
```

::: {.fragment .red-text}
`%let` creates it; `&` uses it. Use **double quotes**: `"&CarMake"` works, `'&CarMake'` does not.
:::

## Macro variables
Change `Kia` to `Toyota` on the first line and run it again.

````sas
%let CarMake = Kia;

proc print data=sashelp.cars;
    where Make = "&CarMake";
    var Type Make Model MSRP;
run;

proc freq data=sashelp.cars;
    where Make = "&CarMake";
    tables Origin;
run;
````

## Exercise

Use `ma505.ow_stats`:

1. For **Player 1**, in the maps **Busan, Nepal, Junkertown, and Oasis**, find:
    a. the average number of eliminations and deaths
    b. the top 5 values for damage
2. Do the same for **Player 2**.

*Hint: a macro variable means you only change one line for Player 2.*

<div class="solution-start"></div>

````sas
%let player = player1;    /* change to player2 and re-run */
%let maps = "Busan", "Nepal", "Junkertown", "Oasis";

/* a. average eliminations and deaths */
proc means data=ma505.ow_stats mean;
    where Player = "&player" and Map in (&maps);
    var Elimination Deaths;
run;

/* b. top 5 damage values: just the Extreme Observations table */
ods select ExtremeObs;
proc univariate data=ma505.ow_stats;
    where Player = "&player" and Map in (&maps);
    var Damage;
run;
````

Output shown for both players (the code above, run once with `player1` and once with `player2`).

- **player1** (7 games on those maps): averages **12.29 eliminations** and **6.86 deaths**; top 5 damage: **10948, 8196, 6301, 5320, 4307**.
- **player2**: averages **16 eliminations** and **7 deaths**; only **3 games** on those maps, so `PROC UNIVARIATE` can list just 3 values: **4602, 3349, 2949**.
