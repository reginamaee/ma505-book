## SAS libraries

A **SAS library** is a collection of SAS files that are stored in the same folder or directory.

Typing the full file path every time is long and easy to get wrong:

```{=html}
<div class="anno"><div class="anno-row"><pre class="anno-code"><span class="kw">proc contents</span> data=<span style="text-decoration:underline;text-decoration-color:#e53935;text-decoration-thickness:3px;text-underline-offset:5px;">"/home/u63936157/ow_hero.sas7bdat"</span>;
<span class="kw">run</span>;</pre></div></div>
```

::: {.fragment}
Instead, we can store our data in a **SAS library** for ease of access.
This is done with the **global** statement `LIBNAME`.
:::

## The `LIBNAME` statement

```{=html}
<div class="anno"><div class="anno-row"><pre class="anno-code"><span class="kw">libname</span> <i>libref</i> "<i>path-to-folder</i>";</pre></div></div>
```

- `libref` is the short name you will use for the library (up to 8 characters, starting with a letter or underscore).
- The path is the **folder** that holds your data, not a single file.
- After this runs, refer to any table in that folder as `libref.table-name`.

## What is a libref?

```{=html}
<div class="anno"><div class="anno-row"><pre class="anno-code"><span class="kw">proc contents</span> data=<span class="libref-circle">ma505</span>.overwatch;
<span class="kw">run</span>;</pre>
<div class="anno-callout fragment"><div class="arrow-left"></div><span class="libref-label">libref</span></div></div></div>
```

::: {.fragment .red-text}
A ***libref*** is:

- a pointer, or shortcut, to existing data
- the libref may be deleted, but the data stays in the same place
- to get the libref back, run the `LIBNAME` statement again
:::

## The work library

- **work** is SAS's scratch space. Every session starts with a fresh, empty one.
- Any table you create **without a libref** goes there: `data dates;` is really `data work.dates;`
- Use it for intermediate steps you don't need to keep.
- Everything in work is **deleted when you sign out**. To keep a table, save it with a permanent libref instead: `data ma505.dates;`

## Your turn: get the Overwatch data into SAS

1. Download `overwatch_heros.csv` from Moodle.
2. In SAS Studio, open **Server Files and Folders**, right-click **Files (Home)** → **Upload Files**, and upload the file.
3. Right-click the uploaded file → **Properties** to see its full path, e.g. `/home/u12345678/overwatch_heros.csv`.
4. Create the **ma505** library, then **import** the CSV into it as a SAS table (use your own user ID):

````sas
/* point the libref ma505 at your home folder */
libname ma505 "/home/u63936157";

/* import the CSV into the library as a SAS table called "overwatch" */
proc import datafile="/home/u63936157/overwatch_heros.csv"
    out=ma505.overwatch
    dbms=csv
    replace;
run;
````

## Viewing column attributes: `PROC CONTENTS`

`PROC CONTENTS` shows a table's **descriptor**: its size, when it was created, and every column's attributes.

````sas
proc contents data=ma505.overwatch;
run;
````

## Reading the output

- The first table is the **descriptor**: table name, number of rows (*Observations*), number of columns (*Variables*), and dates created/modified.
- The second table lists each column's **attributes**: *Type* (`Num` or `Char`), *Len* (length), and its *Format*.
- Notice `Date` is `Num` with the format `MMDDYY10.`. It is stored as a number and *displayed* as a date.

## Dates

SAS lets you type a date, time, or datetime as a **literal**: the value in quotes, followed by `d`, `t`, or `dt`.

No libref here, so `dates` is saved in **work**.

````sas
data dates;
    d  = '01JAN2024'd;          /* date     */
    t  = '18:00't;              /* time     */
    dt = '01JAN2024:00:00'dt;   /* datetime */
run;

proc print data=dates;
run;
````

## The Libraries pane

In SAS Studio, open **Libraries** in the navigation pane, then **My Libraries**.

::: {.incremental}
1. **WORK**: tables you created without a libref, like `example` and `dates`. These disappear when you sign out.
2. **SASHELP**: sample datasets. Double-click **CLASS** to view the table you used on Day 1.
3. **MA505**: appears after you run your `LIBNAME` statement. Click **Refresh** if you don't see it.
:::

## Libraries last for one session

::: {.red-text}
The **libref** is temporary: SAS forgets it when your session ends.
The **data** is permanent: your tables stay in the folder.
Run `LIBNAME` again (or put it in your autoexec) to reconnect.
:::

::: {.fragment}
**So why use libraries?**

- **Short names instead of long paths:** `ma505.overwatch` instead of `"/home/u63936157/overwatch.sas7bdat"`.
- **Your work gets saved:** tables written with a libref (`data ma505.clean;`) are real files that survive sign-out. Tables with no libref go to `work`, which *is* deleted.
- **One place to change:** if the folder moves, edit one `LIBNAME` line, not every PROC.
- **Organization:** separate librefs for separate projects, or for raw vs. cleaned data.
- **One line to restore:** just the `LIBNAME` at the top of your program, or nothing at all with an autoexec.
:::

## Keeping a library: the autoexec file

The **autoexec** file is a SAS program that runs automatically every time you start SAS Studio.
Put your `LIBNAME` statement there and the library is ready every time you log in.

1. In SAS Studio, click **More application options** (top right) → **Edit Autoexec File**.
2. Add your `LIBNAME` statement (example below).
3. Click **Run** and check the log for errors.
4. Click **Save**. Sign out and back in to see **MA505** waiting in the Libraries pane.

````sas
/* autoexec.sas: runs automatically each time SAS Studio starts */
libname ma505 "/home/u63936157";
````
